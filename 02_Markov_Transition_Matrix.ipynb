```python
import numpy as np
from collections import defaultdict

# Example DNA sequence (replace with your own or load from a file)
sequence = "ACGTGCTAGCTAGCTAGCTAGCTAAGCTTGC"

# Ensure uppercase and remove non-ACGT characters
sequence = ''.join([base for base in sequence.upper() if base in 'ACGT'])

# Initialize count matrix
nucleotides = ['A', 'C', 'G', 'T']
counts = defaultdict(lambda: defaultdict(int))

# Count transitions
for i in range(len(sequence) - 1):
    current = sequence[i]
    next_base = sequence[i + 1]
    counts[current][next_base] += 1

# Build transition matrix
transition_matrix = np.zeros((4, 4))
for i, from_base in enumerate(nucleotides):
    total = sum(counts[from_base].values())
    for j, to_base in enumerate(nucleotides):
        if total > 0:
            transition_matrix[i, j] = counts[from_base][to_base] / total

# Print transition matrix with labels
print("Transition Matrix (1st-order Markov model):\n")
print("     " + "  ".join(nucleotides))
for i, from_base in enumerate(nucleotides):
    row = "  ".join(f"{transition_matrix[i, j]:.3f}" for j in range(4))
    print(f"{from_base}  {row}")
```
