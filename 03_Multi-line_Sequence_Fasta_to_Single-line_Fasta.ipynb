```python
def flatten_fasta(input_path, output_path):
    with open(input_path, 'r') as infile, open(output_path, 'w') as outfile:
        current_header = None
        current_sequence = []

        for line in infile:
            line = line.strip()
            if line.startswith(">"):
                if current_header:
                    # Write previous entry
                    outfile.write(current_header + '\n')
                    outfile.write(''.join(current_sequence) + '\n')
                current_header = line
                current_sequence = []
            else:
                current_sequence.append(line)

        # Write the last entry
        if current_header:
            outfile.write(current_header + '\n')
            outfile.write(''.join(current_sequence) + '\n')

# Run on the uploaded file
flatten_fasta("multiline_input.fasta", "flattened_output.fasta")
print("Flattened FASTA saved to: flattened_output.fasta")
```
